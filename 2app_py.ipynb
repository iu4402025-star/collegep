```python

import streamlit as st
import pandas as pd

# ---------------- PAGE CONFIG ----------------
st.set_page_config(
    page_title="STYLEHUB | Fashion Store",
    page_icon="👗",
    layout="wide"
)

# ---------------- CUSTOM CSS ----------------
st.markdown("""
<style>
@import url('https://fonts.googleapis.com/css2?family=Poppins:wght@300;400;500;600;700&display=swap');

html, body, [class*="css"] {
    font-family: 'Poppins', sans-serif;
}

.stApp {
    background: #f8f6f2;
}

[data-testid="stHeader"] {
    background: transparent;
}

.hero {
    background: linear-gradient(110deg, #171717, #594638);
    padding: 55px 35px;
    border-radius: 18px;
    color: white;
    margin-bottom: 25px;
}

.hero h1 {
    font-size: 48px;
    font-weight: 700;
    color: white;
}

.hero p {
    font-size: 17px;
    color: #eee5dc;
}

.product-card {
    background: white;
    padding: 13px;
    border-radius: 14px;
    margin-bottom: 15px;
    border: 1px solid #eee;
}

.product-name {
    font-size: 17px;
    font-weight: 600;
    color: #242424;
}

.product-price {
    font-size: 18px;
    font-weight: 700;
    color: #9a643c;
}

.section-title {
    font-size: 28px;
    font-weight: 700;
    color: #28221e;
    margin-top: 25px;
}

.stButton button {
    background: #29231e;
    color: white;
    border-radius: 8px;
    border: none;
}

.stButton button:hover {
    background: #9a643c;
    color: white;
}
</style>
""", unsafe_allow_html=True)


# ---------------- SHOP INFORMATION ----------------

SHOP_NAME = "STYLEHUB"

products = [
    {
        "id": 1,
        "name": "Premium Men's T-Shirt",
        "category": "Men",
        "price": 1800,
        "image": "https://images.unsplash.com/photo-1521572163474-6864f9cf17ab?w=600",
        "description": "Comfortable cotton T-shirt"
    },
    {
        "id": 2,
        "name": "Classic Men's Jacket",
        "category": "Men",
        "price": 4500,
        "image": "https://images.unsplash.com/photo-1551028719-00167b16eac5?w=600",
        "description": "Stylish everyday jacket"
    },
    {
        "id": 3,
        "name": "Women's Fashion Dress",
        "category": "Women",
        "price": 3500,
        "image": "https://images.unsplash.com/photo-1595777457583-95e059d581b8?w=600",
        "description": "Elegant modern dress"
    },
    {
        "id": 4,
        "name": "Women's Casual Outfit",
        "category": "Women",
        "price": 2800,
        "image": "https://images.unsplash.com/photo-1483985988355-763728e1935b?w=600",
        "description": "Beautiful casual fashion"
    },
    {
        "id": 5,
        "name": "Kids Fashion Outfit",
        "category": "Kids",
        "price": 1600,
        "image": "https://images.unsplash.com/photo-1519238263530-99bdd11df2ea?w=600",
        "description": "Comfortable kids clothing"
    },
    {
        "id": 6,
        "name": "Premium Hoodie",
        "category": "Men",
        "price": 3200,
        "image": "https://images.unsplash.com/photo-1556821840-3a63f95609a7?w=600",
        "description": "Warm and trendy hoodie"
    },
    {
        "id": 7,
        "name": "Women's Summer Wear",
        "category": "Women",
        "price": 2400,
        "image": "https://images.unsplash.com/photo-1515886657613-9f3515b0c78f?w=600",
        "description": "Light summer collection"
    },
    {
        "id": 8,
        "name": "Kids Casual Wear",
        "category": "Kids",
        "price": 1900,
        "image": "https://images.unsplash.com/photo-1503919005314-30d93d07d823?w=600",
        "description": "Trendy everyday kids wear"
    }
]


# ---------------- SESSION STATE ----------------

if "cart" not in st.session_state:
    st.session_state.cart = {}

if "page" not in st.session_state:
    st.session_state.page = "Home"


# ---------------- HEADER ----------------

col1, col2 = st.columns([2, 1])

with col1:
    st.markdown(
        "<h2 style='color:#29231e;'>👗 STYLEHUB</h2>"
        "<p style='color:#92765e;'>YOUR STYLE, YOUR IDENTITY</p>",
        unsafe_allow_html=True
    )

with col2:
    st.markdown(
        f"<h4 style='text-align:right;color:#29231e;'>🛒 "
        f"Cart: {sum(st.session_state.cart.values())} items</h4>",
        unsafe_allow_html=True
    )

st.divider()


# ---------------- NAVIGATION ----------------

nav1, nav2, nav3, nav4 = st.columns(4)

with nav1:
    if st.button("🏠 Home", use_container_width=True):
        st.session_state.page = "Home"

with nav2:
    if st.button("👕 Shop", use_container_width=True):
        st.session_state.page = "Shop"

with nav3:
    if st.button("🛒 My Cart", use_container_width=True):
        st.session_state.page = "Cart"

with nav4:
    if st.button("📞 Contact", use_container_width=True):
        st.session_state.page = "Contact"


st.divider()


# ---------------- HOME PAGE ----------------

if st.session_state.page == "Home":

    st.markdown("""
    <div class="hero">
        <h1>NEW FASHION COLLECTION</h1>
        <p>Discover your perfect style with our latest clothing collection.</p>
        <h3>STYLE THAT SPEAKS FOR YOU.</h3>
    </div>
    """, unsafe_allow_html=True)

    st.markdown(
        "<div class='section-title'>✨ Featured Collection</div>",
        unsafe_allow_html=True
    )

    selected_products = products[:4]

    cols = st.columns(4)

    for col, product in zip(cols, selected_products):
        with col:
            st.image(product["image"], use_container_width=True)
            st.markdown(
                f"<div class='product-name'>{product['name']}</div>",
                unsafe_allow_html=True
            )
            st.markdown(
                f"<div class='product-price'>Rs. {product['price']:,}</div>",
                unsafe_allow_html=True
            )

            if st.button(
                "Add to Cart",
                key=f"home_{product['id']}",
                use_container_width=True
            ):
                pid = product["id"]
                st.session_state.cart[pid] = (
                    st.session_state.cart.get(pid, 0) + 1
                )
                st.success("Added to cart!")
                st.rerun()

    st.markdown("---")
    st.markdown("### 🚚 Why Shop With Us?")
    a, b, c = st.columns(3)

    a.info("📦 Quality Products\n\nCarefully selected clothing.")
    b.info("💰 Affordable Prices\n\nFashion for every budget.")
    c.info("❤️ Customer Service\n\nWe care about your satisfaction.")


# ---------------- SHOP PAGE ----------------

elif st.session_state.page == "Shop":

    st.title("🛍️ Our Clothing Collection")
    st.write("Explore the latest fashion products.")

    search = st.text_input("🔍 Search clothing products")

    category = st.selectbox(
        "Select Category",
        ["All", "Men", "Women", "Kids"]
    )

    filtered = products

    if category != "All":
        filtered = [
            p for p in filtered if p["category"] == category
        ]

    if search:
        filtered = [
            p for p in filtered
            if search.lower() in p["name"].lower()
        ]

    if not filtered:
        st.warning("No products found.")

    cols = st.columns(4)

    for index, product in enumerate(filtered):
        with cols[index % 4]:

            st.image(product["image"], use_container_width=True)

            st.markdown(
                f"<div class='product-name'>{product['name']}</div>",
                unsafe_allow_html=True
            )

            st.caption(product["description"])

            st.markdown(
                f"<div class='product-price'>Rs. {product['price']:,}</div>",
                unsafe_allow_html=True
            )

            if st.button(
                "Add to Cart 🛒",
                key=f"shop_{product['id']}",
                use_container_width=True
            ):
                pid = product["id"]
                st.session_state.cart[pid] = (
                    st.session_state.cart.get(pid, 0) + 1
                )
                st.success("Product added!")
                st.rerun()


# ---------------- CART PAGE ----------------

elif st.session_state.page == "Cart":

    st.title("🛒 Your Shopping Cart")

    total = 0

    if not st.session_state.cart:
        st.info("Your cart is empty. Visit our shop to add products.")

    else:
        for pid in list(st.session_state.cart.keys()):

            product = next(
                p for p in products if p["id"] == pid
            )

            qty = st.session_state.cart[pid]
            subtotal = product["price"] * qty
            total += subtotal

            c1, c2, c3 = st.columns([3, 1, 1])

            with c1:
                st.write(f"**{product['name']}**")
                st.caption(f"Rs. {product['price']:,} each")

            with c2:
                st.write(f"Quantity: {qty}")

            with c3:
                if st.button("Remove", key=f"remove_{pid}"):
                    del st.session_state.cart[pid]
                    st.rerun()

            st.write(f"Subtotal: Rs. {subtotal:,}")
            st.divider()

        st.subheader(f"Total Amount: Rs. {total:,}")

        customer_name = st.text_input("Customer Name")
        phone = st.text_input("Phone Number")
        address = st.text_area("Delivery Address")

        if st.button("Place Order", use_container_width=True):

            if customer_name and phone and address:
                st.success("Order details received!")
                st.write("Customer:", customer_name)
                st.write("Phone:", phone)
                st.write("Address:", address)
                st.write("Order Total:", f"Rs. {total:,}")
                st.info(
                    "Please contact the shop owner to confirm "
                    "payment and delivery."
                )
            else:
                st.error("Please complete all customer details.")


# ---------------- CONTACT PAGE ----------------

elif st.session_state.page == "Contact":

    st.title("📞 Contact STYLEHUB")

    st.markdown("""
    ### We would love to hear from you!

    **Business:** STYLEHUB Fashion Store

    **Address:** Add your shop address here.

    **Phone:** +92 300 1234567

    **Email:** stylehub@example.com

    **Opening Hours:** Monday – Saturday
    """)

    st.markdown(
        "[💬 Contact Us on WhatsApp](https://wa.me/923001234567)"
    )

    st.write("Follow us on social media for new collections!")


# ---------------- FOOTER ----------------

st.divider()

st.markdown("""
<div style="text-align:center; padding:20px; color:#786b60;">
    <h3>STYLEHUB 👗</h3>
    Your Fashion Destination
    <p>© 2026 STYLEHUB. All Rights Reserved.</p>
</div>
""", unsafe_allow_html=True)
```
